```python
%%writefile calc.py

def add(a, b):
    return a + b

def safe_divide(a, b):
    if b == 0:
        raise ValueError("b cannot be zero")
    return a / b

def chunk_text(text, size, overlap):
    if size <= overlap:
        raise ValueError("size must be greater than overlap")
    out, i = [], 0
    while i < len(text):
        out.append(text[i:i + size])
        i += size - overlap
    return out
```
